```python
import keras
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
df = load_iris()
X, y = df.data, df.target
```

```python
X_train, X_test,y_train, y_test = train_test_split(X, y, test_size=0.2)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.fit_transform(X_test)
```

```python
X_train_scaled = keras.ops.convert_to_tensor(X_train_scaled)
X_test_scaled = keras.ops.convert_to_tensor(X_test_scaled)
```

```python
y_train = keras.ops.convert_to_tensor(y_train)
```

```python
_, input_shape=X_train_scaled.shape
```

```python
inputs = keras.Input(shape=(input_shape,))
x = keras.layers.Dense(8, activation='relu')(inputs)
x = keras.layers.Dense(16, activation='relu')(x)
output = keras.layers.Dense(3, activation='softmax')(x)

Model = keras.Model(inputs=inputs, outputs=output)
```

```python
Model.summary()
```

```python
Model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
Model.fit(
    X_train_scaled, y_train,epochs=10, batch_size=10
)
```
